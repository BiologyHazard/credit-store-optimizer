```python
# K_·_813_·
C = 813
for n in range(11):
    plt.plot([K···(n, C, c) for c in range(301)], label=f'K_{n}_813_·')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.title('K_·_813_·')
plt.xlabel('Carried credit c')
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# K_10_813_·
n = 10
C = 813
plt.plot([0, 300], [K···(n, C, 0), K···(n, C, 300)], label=f'Line segment connecting endpoints', linestyle='--', color='#AAA', linewidth=1)
plt.plot([K···(n, C, c) for c in range(301)], label=f'K_{n}_813_·')
plt.legend()
plt.title('K_10_813_·')
plt.xlabel('Carried credit c')
# plt.xlim(0, 300)
# plt.ylim(840, 856)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 标准化的 K_·_813_·
C = 813
for n in range(11):
    plt.plot([K···(n, C, c) - K···(n, C, 0) for c in range(301)], label=f'Normalized K_{n}_813_·', linewidth=1)
plt.legend()
plt.title('Normalized K_·_813_·')
plt.xlabel('Carried credit c')
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 每日获得的信用的价值
plt.plot(每日获得的信用的价值(range(1001)), label='Value of daily credit')
plt.legend()
plt.title('Value of daily credit with daily credit income')
plt.xlabel('Daily credit income')
plt.ylabel('Sanity value')
plt.xlim(0, 1000)
plt.ylim(0, 100)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# “1”信用的价值
plt.plot(每日获得的信用的价值的导数(range(1001)), label='Value of "1" credit')
plt.legend()
plt.title('Value of "1" credit (the curve has been smoothed)')
plt.xlabel('Daily credit income')
plt.ylabel('Sanity value')
plt.xlim(-10, 1000)
plt.ylim(bottom=0)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 信用能购买的商品碎片的价值
plt.plot(信用能购买的商品碎片的价值(range(1501)), label='Value of item fragments that credit can buy')
plt.legend()
plt.title('Value of item fragments that credit can buy')
plt.xlabel('Credit c')
plt.ylabel('Sanity value')
plt.xlim(0, 1200)
plt.ylim(0, 100)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 最优策略的近似
plt.plot(信用能购买的商品碎片的价值(range(1501)), label='Value of item fragments that credit can buy (as an approximation of K_N-1_·_·)', linewidth=1)
for C in range(200, 1001, 200):
    plt.plot(
        range(C - 150, C + 151),
        [K_N减1_·_·(C, c) - K_N减1_·_·(C, 150) + 信用能购买的商品碎片的价值(C) for c in range(301)],
        label=f'K_{N - 1}_{C}_· (suitably translated)',
        linewidth=1,
    )
plt.legend()
plt.title('The real value and approximation of K_10_·_·')
plt.xlabel('Credit c')
plt.ylabel('Sanity value')
plt.xlim(0, 1200)
plt.ylim(0, 100)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 信用能购买的商品碎片的价值和每日获得的信用的价值
plt.plot(信用能购买的商品碎片的价值(range(1501)), label='Value of item fragments that credit can buy')
plt.plot(每日获得的信用的价值(range(1001)), label='Value of daily credit')
plt.legend()
plt.title('Value of item fragments that credit can buy and value of daily credit')
plt.xlabel('Credit c')
plt.ylabel('Sanity value')
plt.xlim(0, 1200)
plt.ylim(0, 100)
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```

```python
# 性价比阈值的取法
plt.plot([(K_N减1_·_·(C, 300) - K_N减1_·_·(C, 0)) / 300 for C in range(1001)], label=f'(K_{N - 1}_·_300 - K_{N - 1}_·_0) / 300')
plt.plot([信用能购买的商品碎片的价值(c + 1) - 信用能购买的商品碎片的价值(c) for c in range(1500)], label='Derivative of value of item fragments that credit can buy')
plt.title('The threshold of efficiency')
plt.xlabel('Daily credit income C')
plt.ylabel('Threshold of efficiency (Sanity per credit)')
plt.xlim(-10, 1000)
plt.ylim(bottom=0)
plt.legend()
plt.minorticks_on()
plt.grid(color='lightgray', which='both')
```
