```python
import random
from collections import Counter

from 信用交易所商品信息 import 信用交易所商品名称列表
from 信用交易所模型 import 信用交易所
from 信用交易所购买策略 import 信用交易所购买策略, 按性价比从高到低购买_并设定性价比阈值策略, 计算总价值, 计算总价格
from 统计结果 import 信用交易所统计结果


def 模拟一轮购买(
    策略: 信用交易所购买策略,
    每日获得的信用: int,
    初始继承信用: int,
    商店列表: list[信用交易所],
) -> Counter[str]:
    当前信用 = 初始继承信用 + 每日获得的信用
    已购买的商品 = Counter()
    已购买总价值 = 0
    for 商店 in 商店列表:
        购买指标向量 = 策略.求解(信用交易所(当前信用, 商店.商品列表))
        当前信用 -= 计算总价格(商店, 购买指标向量)
        if 当前信用 < 0:
            raise ValueError('信用不足')
        已购买总价值 += 计算总价值(商店, 购买指标向量)
        for 商品, 购买指标 in zip(商店.商品列表, 购买指标向量):
            已购买的商品[商品.物品名称] += int(购买指标)
        继承的信用 = min(当前信用, 300)
        当前信用 = 继承的信用 + 每日获得的信用
    return 已购买的商品


每日获得的信用 = 942
初始继承信用 = 0

random.seed(0)
顺序 = list(range(len(信用交易所统计结果))) * 5
random.shuffle(顺序)
商店列表 = [信用交易所统计结果[商店序号] for 商店序号 in 顺序]

已购买的商品 = 模拟一轮购买(
    按性价比从高到低购买_并设定性价比阈值策略(每日获得的信用=每日获得的信用),
    每日获得的信用,
    初始继承信用,
    商店列表,
)
平均每日购买的商品 = {商品名称: 已购买的商品[商品名称] / len(商店列表) for 商品名称 in 信用交易所商品名称列表}
# 平均每日购买的商品 = dict(sorted(平均每日购买的商品.items(), key=lambda x: x[1], reverse=True))
一年的天数 = 365
平均每年购买的商品 = {商品名称: 数量 * 一年的天数 for 商品名称, 数量 in 平均每日购买的商品.items()}
```

```python
import pandas as pd

# 平均每日购买的商品
df = pd.DataFrame(平均每日购买的商品.items(), columns=['商品', '数量'])
# 把“商品”列设置为索引
df.set_index('商品', inplace=True)


df.to_clipboard()
```
